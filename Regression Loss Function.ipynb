```python
#Loss Function
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
```

```python
# ============================================================
# 1. MEAN SQUARED ERROR (MSE)
# ============================================================
#
# Formula:
#
# MSE = (1/n) * Σ(y_true - y_pred)^2
#
# Mainly used for:
# - Regression problems
#
# Example:
# Predicting house prices, temperature, sales, etc.
# ============================================================

print("\n" + "=" * 60)
print("1. MEAN SQUARED ERROR (MSE)")
print("=" * 60)

y_true = torch.tensor([10.0, 20.0, 30.0])
y_pred = torch.tensor([12.0, 18.0, 29.0])

mse_loss = nn.MSELoss()
mse = mse_loss(y_pred, y_true)

print("Actual values:   ", y_true)
print("Predicted values:", y_pred)
print("MSE Loss:", mse.item())
                                                                                                         
print("""
Interpretation:
MSE calculates the squared difference between actual and
predicted values.

Large errors are penalized heavily because the error is squared.
""")
```

```python
# ============================================================
# 2. MEAN ABSOLUTE ERROR (MAE)
# ============================================================
#
# Formula:
#
# MAE = (1/n) * Σ|y_true - y_pred|
#
# Mainly used for:
# - Regression
#
# Compared with MSE, MAE is less sensitive to outliers.
# ============================================================

print("\n" + "=" * 60)
print("2. MEAN ABSOLUTE ERROR (MAE)")
print("=" * 60)

mae_loss = nn.L1Loss()
mae = mae_loss(y_pred, y_true)

print("MAE Loss:", mae.item())

print("""
Interpretation:
MAE calculates the absolute difference between actual and
predicted values.

Unlike MSE, it does not square the error, so extreme errors
have less influence.
""")
```

```python
# ============================================================
# 3. BINARY CROSS-ENTROPY (BCE)
# ============================================================
#
# Used for binary classification.
#
# Example:
#   0 -> Cat
#   1 -> Dog
#
# The model outputs a probability between 0 and 1.
#
# Formula:
#
# BCE = -[y log(p) + (1-y) log(1-p)]
#
# In PyTorch, BCEWithLogitsLoss is preferred because it combines
# sigmoid + BCE in a numerically stable way.
# ============================================================

print("\n" + "=" * 60)
print("3. BINARY CROSS-ENTROPY")
print("=" * 60)

# Raw model outputs (logits)
logits = torch.tensor([2.0, -1.0, 0.5])

# Actual labels
labels = torch.tensor([1.0, 0.0, 1.0])

bce_loss = nn.BCEWithLogitsLoss()
bce = bce_loss(logits, labels)

print("Logits:", logits)
print("Labels:", labels)
print("BCE Loss:", bce.item())

probabilities = torch.sigmoid(logits)

print("Predicted probabilities:", probabilities)

print("""
Interpretation:
BCE is commonly used for binary classification.

For example:

0 -> Negative class
1 -> Positive class

A prediction close to the correct class receives a small loss.
A confident but incorrect prediction receives a very large loss.
""")

                                                                                
```

```python
# ============================================================
# 7. HUBER LOSS
# ============================================================
#
# Huber Loss combines properties of MSE and MAE.
#
# For small errors:
#
#       behaves like MSE
#
# For large errors:
#
#       behaves more like MAE
#
# This makes it useful when we want some robustness to outliers.
# ============================================================

print("\n" + "=" * 60)
print("7. HUBER LOSS")
print("=" * 60)

huber_loss = nn.HuberLoss()

huber = huber_loss(y_pred, y_true)

print("Huber Loss:", huber.item())

print("""
Interpretation:

Small errors:
    Huber behaves approximately like MSE.

Large errors:
    Huber behaves approximately like MAE.

Therefore, Huber Loss is often useful when a regression
dataset contains outliers.
""")
```

```python
# ============================================================
# 4. CATEGORICAL CROSS-ENTROPY
# ============================================================
#
# Used for multi-class classification.
#
# Example:
#
# Class 0 = Cat
# Class 1 = Dog
# Class 2 = Bird
#
# The model produces one score (logit) for every class.
#
# CrossEntropyLoss internally performs:
#
# Softmax + Negative Log Likelihood
#
# Therefore, DO NOT apply softmax before passing logits to
# nn.CrossEntropyLoss().
# ============================================================

print("\n" + "=" * 60)
print("4. CATEGORICAL CROSS-ENTROPY")
print("=" * 60)

# Three samples, three classes
logits = torch.tensor([
    [2.0, 0.5, 0.1],
    [0.2, 2.5, 0.3],
    [0.1, 0.2, 3.0]
])

# Correct class for each sample
targets = torch.tensor([0, 1, 2])

cross_entropy = nn.CrossEntropyLoss()
ce_loss = cross_entropy(logits, targets)

print("Logits:")
print(logits)

print("\nCorrect classes:")
print(targets)

print("\nCross-Entropy Loss:", ce_loss.item())

probabilities = F.softmax(logits, dim=1)

print("\nClass probabilities:")
print(probabilities)

print("""
Interpretation:

Each row represents one sample.

For example:

[2.0, 0.5, 0.1]

means the model gives different scores to:

Class 0 -> 2.0
Class 1 -> 0.5
Class 2 -> 0.1

Softmax converts these scores into probabilities.

CrossEntropyLoss compares those probabilities with the
correct class.
""")

```

```python
# ============================================================
# 6. HINGE LOSS
# ============================================================
#
# Traditionally associated with Support Vector Machines (SVMs),
# but it can also be used in neural networks.
#
# Formula:
#
# L = max(0, 1 - y * prediction)
#
# where:
#
# y ∈ {-1, +1}
#
# Hinge loss encourages the model not only to classify correctly
# but also to classify with a margin.
# ============================================================

print("\n" + "=" * 60)
print("6. HINGE LOSS")
print("=" * 60)

hinge_loss = nn.HingeEmbeddingLoss()

# Example for demonstration
inputs = torch.tensor([1.5, -0.5, 2.0])
targets_hinge = torch.tensor([1.0, -1.0, 1.0])

hinge = hinge_loss(inputs, targets_hinge)

print("Inputs:", inputs)
print("Targets:", targets_hinge)
print("Hinge Loss:", hinge.item())

print("""
Important:
Hinge loss is commonly associated with SVM-style classification.

The goal is to create a margin between classes rather than
simply getting the class label correct.
""")
```
